```python
with open("file2.txt","r") as f:
    data=f.read().split()
    data_lower=[i.lower() for i in data]
    sort_data=sorted(data_lower)
with open("new.txt","w") as f2:
    for i in sort_data:
        f2.write(i+"\n")
```

```python
with open("new.txt","r") as f:
    lines=f.readlines()
    print(len(lines))
    words=0
    chars=0
    for i in lines:
        words=words+len(i.split())
        chars=chars+len(i)
    print(words,chars)
```
