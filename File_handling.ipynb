```python
#write
f=open("data.txt","w")
content = "i am a file"
f.write(content)
f.writelines(["\n3rd line","\n4th line"])#writelines
print("file created")
f.close()
```

```python
#built in method
f.name
```

```python
#built in method
f.mode
```

```python
#read
with open("data.txt","r") as file:
  content=file.read(10)
  print(content)
  print(file.tell())
  file.seek(2)
  print(file.read())
```

```python
#read lines
with open("data.txt","r") as file:
  content=file.readlines()
  print(content)
```

```python
#append
f=open("data.txt","a+")
content="\nrevathi"
f.write(content)
print("appended")
f.close()
```
