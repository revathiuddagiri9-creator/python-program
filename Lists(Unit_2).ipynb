```python
L1=[1,2,3,55,66,88,99.0,"hello","name"]
L2=[100,220,440]

print(L1+L2)       # concatenation
print(L2*3)        # repetition
print(L1[5])       # indexing
print(100 in L2)   # membership operator
print(L1[2:7])     # slicing
print(L1[::-1])    # reverse order
print(L1[-7:-1])

for i in L2:
    print(i)       # iteration through a list
```
