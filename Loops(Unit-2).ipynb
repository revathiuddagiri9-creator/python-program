```python
# even numbers in reverse order
n=int(input("enter:"))
i=n
while(i>=1):
    if(i%2==0):
        print(i,end=" ")
    i=i-1
```

```python
#for loop in python
n=int(input())
for i in range(n,1,-1):
    print(i,end=" ")
```

```python
#multiplication table for a given num
n=int(input())
for i in range(1,11):
  print(f"{n}*{i}={n*i}")
```

```python
#given number is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("not a prime")
    break
else:
          print("it is a prime")
```

```python
#prime numbers between two numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
      print(i)
```

```python
#sum of digits 0f given two numbers
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

```python
#print 1 to 10 numbers 10 times
for i in range(1,11):
  for j in range(1,11):
    print(j,end="")
  print()
```

```python
# sequence find 1-2+3-4+5-6...n
n=int(input())
sum=0
for i in range(1,n+1):
    if(i%2==0):
        sum=sum-i
    else:
        sum=sum+i
print(sum)
```
