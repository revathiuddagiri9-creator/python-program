```python
#given number is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("not a prime")
    break
else:
          print("it is a prime")
```

```python
#prime numbers between two numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
      print(i)
```

```python
#sum of digits 0f given two numbers
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
