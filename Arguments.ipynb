```python
#palindrome
def palindrome(num):
  original = num
  rev = 0
  while(num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  if(rev==original):
    return True
  else:
    return False
print(palindrome (121))
print(palindrome (123))
```

```python
#position arguments
def sub(a,b):
  c=a-b
  print(c)
x=int(input())
y=int(input())
sub(x,y)
```

```python
#reverse the string
def reverse_str(s):
  rev=s[::-1]
  return rev
print(reverse_str("revathi"))
#without above logic
def reverse_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(reverse_str("daddy"))
```

```python
#default arguments
def print_msg(name,branch="AI&DS"):
  print("the name is",name,"the branch is",branch)
print_msg("sai")
```

```python
#keyword arguments
def print_msg(name,branch):
  print("the name is",name,"the branch is",branch)
print_msg(branch="AI&DS",name="sai")
```

```python
#variable length arguments
def add(*num):
  print(num)
  print(sum(num))
add(1,2,3,4,5,6)
```

```python
#function that returns multiple
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is{res_sum},count is{res_count},the average is{res_avg}")
```
