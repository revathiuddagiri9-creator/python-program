```python
#sets
s={1,2,3,"hello",1,2,3}
print(s)
a={}
print(type(a))
a=set()
print(type(a))
```

```python
#1.union 2.intersection 3.difference 4.symmetric difference
A={1,2,3}
B={3,4,5}
print(A|B)  #union
print(A&B)  #intersection
print(A-B)  #difference
print(A^B)  #symmetric difference
```

```python
L1=[12,23,44]
L=[23,14,64]
A=set(L)
B=set(L1)
print(A)
print(A&B)
```

```python
#2222 is yes else no
n=int(input())
s=str(n)
print(s)
m=set(s)
if(len(m)>1):
  print("no")
else:
  print("yes")
print(m)
```
