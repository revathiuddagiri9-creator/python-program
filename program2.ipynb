```python
#finding greatest among 3 numbers
a=int(input("enter a value"))
b=int(input("enter b value"))
c=int(input("enter c value"))
if(a>b and a>c):
  print(f"{a} is greater")
elif(b>c and b>a):
  print(f"{b} is greater")
else:
  print(f"{c} is greater")
```

```python
#print FIZZ if a number is divisible by 3,print BUZZ if a number is divisible by 5 print FIZZBUZZ if it is divisible by both 3 and 5
a=int(input("enter a value"))
if(a%3==0 and a%5!=0):
  print("FIZZ")
if(a%5==0 and a%3!=0):
  print("BUZZ")
if(a%3==0 and a%5==0):
  print("FIZZBUZZ")

```

```python
#program on "match"
day=int(input("enter any number"))
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid")
```
